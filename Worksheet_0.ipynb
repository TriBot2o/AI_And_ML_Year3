```python
def extract_every_other(lst):
    """
    Extracts every other element from a list starting from the first element.

    Parameters:
    lst (list): The original list of elements.

    Returns:
    list: A new list containing elements at indices 0, 2, 4, etc.
    """
    # [::2] means start at the beginning, go to the end, with a step of 2
    return lst[::2]

original_list = [1, 2, 3, 4, 5, 6]
result = extract_every_other(original_list)

print(f"Original List: {original_list}")
print(f"Extracted List: {result}")
```

```python
def get_sublist(lst, start, end):
    """
    Extracts a sublist from a given list from start to end index (inclusive).

    Parameters:
    lst (list): The original list of elements.
    start (int): The starting index.
    end (int): The ending index.

    Returns:
    list: A new list containing elements from start index to end index.
    """
    # We use end + 1 because Python slices are exclusive of the stop index
    return lst[start:end+1]

my_list = [1, 2, 3, 4, 5, 6]
start_idx = 2
end_idx = 4

result = get_sublist(my_list, start_idx, end_idx)
print(f"Sublist from index {start_idx} to {end_idx}: {result}")
```

```python
def reverse_list(lst):
    """
    Reverses a list using slicing.

    Parameters:
    lst (list): The original list of elements.

    Returns:
    list: A new list with the elements in reverse order.
    """
    # [::-1] means start from the end and move backwards with a step of -1
    return lst[::-1]

my_list = [1, 2, 3, 4, 5]
reversed_result = reverse_list(my_list)

print(f"Original: {my_list}")
print(f"Reversed: {reversed_result}")
```

```python
def remove_first_last(lst):
    """
    Removes the first and last elements of a list using slicing.

    Parameters:
    lst (list): The original list of elements.

    Returns:
    list: A new list excluding the first and last elements.
    """
    # Slice from index 1 up to (but not including) the last index -1
    return lst[1:-1]

my_list = [1, 2, 3, 4, 5]
trimmed_list = remove_first_last(my_list)

print(f"Original: {my_list}")
print(f"Trimmed: {trimmed_list}")
```

```python
def get_first_n(lst, n):
    """
    Extracts the first n elements from a list using slicing.

    Parameters:
    lst (list): The original list of elements.
    n (int): The number of elements to extract from the start.

    Returns:
    list: A new list containing the first n elements.
    """
    # Slice from the start up to index n (exclusive)
    return lst[:n]

my_list = [1, 2, 3, 4, 5]
n_value = 3
result = get_first_n(my_list, n_value)

print(f"Original: {my_list}")
print(f"First {n_value} elements: {result}")
```

```python
def get_last_n(lst, n):
    """
    Extracts the last n elements from a list using negative slicing.

    Parameters:
    lst (list): The original list of elements.
    n (int): The number of elements to extract from the end.

    Returns:
    list: A new list containing the last n elements.
    """
    # Start at -n (n positions from the end) and go to the end
    return lst[-n:]

my_list = [1, 2, 3, 4, 5]
n_value = 2
result = get_last_n(my_list, n_value)

print(f"Original: {my_list}")
print(f"Last {n_value} elements: {result}")
```

```python
def reverse_skip(lst):
    """
    Extracts every second element starting from the second-to-last, moving backward.

    Parameters:
    lst (list): The original list of elements.

    Returns:
    list: A new list starting from index -2, moving backward by 2.
    """
    # Start at -2 (second-to-last), stop at the beginning, step of -2
    return lst[-2::-2]

my_list = [1, 2, 3, 4, 5, 6]
result = reverse_skip(my_list)

print(f"Original: {my_list}")
print(f"Result: {result}")
```

### Exercise on Nested List

```python
def flatten(lst):
    """
    Flattens a 2D nested list into a single-dimensional list.

    Parameters:
    lst (list): A list containing other lists as elements.

    Returns:
    list: A single list containing all elements from the sublists.
    """
    # This reads as: "take every 'item' inside every 'sublist' found in 'lst'"
    return [item for sublist in lst for item in sublist]

nested_list = [[1, 2], [3, 4], [5]]
flat_result = flatten(nested_list)

print(f"Original Nested List: {nested_list}")
print(f"Flattened List: {flat_result}")
```

```python
def access_nested_element(lst, indices):
    """
    Accesses a specific element in a nested list using a list of indices.

    Parameters:
    lst (list): The nested list (e.g., a list of lists).
    indices (list): A list of integers representing the path to the element.

    Returns:
    The element located at the specified indices.
    """
    # Start with the full nested list
    current_level = lst

    # Traverse through each index in the provided list
    for index in indices:
        current_level = current_level[index]

    return current_level

my_nested_list = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
my_indices = [1, 2]

result = access_nested_element(my_nested_list, my_indices)
print(f"The element at indices {my_indices} is: {result}")
```

```python
def sum_nested(lst):
    """
    Calculates the sum of all numeric elements in a nested list of any depth.

    Parameters:
    lst (list): A nested list containing numbers or further nested lists.

    Returns:
    float/int: The total sum of all numbers found within the structure.
    """
    total = 0
    for item in lst:
        # Check if the item is a list itself
        if isinstance(item, list):
            # If it's a list, call the function again (Recursion)
            total += sum_nested(item)
        else:
            # If it's a number, add it to the total
            total += item
    return total

nested_data = [[1, 2], [3, [4, 5]], 6]
result = sum_nested(nested_data)

print(f"The total sum of {nested_data} is: {result}")
```

```python
def remove_element(lst, elem):
    """
    Removes all occurrences of a specific element from each sublist in a nested list.

    Parameters:
    lst (list): A nested list (list of lists).
    elem: The specific value to be removed from the sublists.

    Returns:
    list: A new nested list with the specified element removed.
    """
    # Outer comprehension goes through each sublist
    # Inner comprehension keeps 'x' only if it is NOT the element we want to remove
    return [[x for x in sublist if x != elem] for sublist in lst]

my_nested_list = [[1, 2], [3, 2], [4, 5]]
to_remove = 2

result = remove_element(my_nested_list, to_remove)
print(f"Original: {my_nested_list}")
print(f"Modified: {result}")
```

```python
def find_max_nested(lst):
    """
    Finds the maximum value in a nested list of any depth.

    Parameters:
    lst (list): A nested list containing numbers or further nested lists.

    Returns:
    float/int: The highest numeric value found within the structure.
    """
    # Initialize max_val as None or the first numeric element found
    max_val = float('-inf')  # Start with the smallest possible number

    for item in lst:
        if isinstance(item, list):
            # If it's a list, find the max of THAT list recursively
            inner_max = find_max_nested(item)
            if inner_max > max_val:
                max_val = inner_max
        else:
            # If it's a number, compare it to our current max_val
            if item > max_val:
                max_val = item

    return max_val

nested_data = [[1, 2], [3, [4, 5]], 6]
result = find_max_nested(nested_data)

print(f"The maximum element in {nested_data} is: {result}")
```

```python
def count_occurrences(lst, elem):
    """
    Counts how many times a specific element appears in a nested list (2D).

    Parameters:
    lst (list): A nested list containing sublists.
    elem: The specific value to be counted.

    Returns:
    int: The total number of times the element appears across all sublists.
    """
    # Count elem in each sublist and create a list of those counts, then sum them
    return sum(sublist.count(elem) for sublist in lst)

my_nested_list = [[1, 2], [2, 3], [2, 4]]
target_value = 2

result = count_occurrences(my_nested_list, target_value)
print(f"The element {target_value} appears {result} times.")
```

```python
def deep_flatten(lst):
    """
    Flattens a deeply nested list into a single-dimensional list.

    Parameters:
    lst (list): A list that can contain numbers or other nested lists.

    Returns:
    list: A single flat list containing all individual elements.
    """
    flat_list = []

    for item in lst:
        if isinstance(item, list):
            # If the item is a list, recursively flatten it and extend our result
            flat_list.extend(deep_flatten(item))
        else:
            # If the item is not a list (it's a number), just add it
            flat_list.append(item)

    return flat_list

nested_data = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
result = deep_flatten(nested_data)

print(f"Original: {nested_data}")
print(f"Deeply Flattened: {result}")
```

```python
def average_nested(lst):
    """
    Calculates the average of all elements in a 2D nested list.

    Parameters:
    lst (list): A nested list containing sublists of numbers.

    Returns:
    float: The arithmetic mean of all numbers in the nested structure.
    """
    # Step 1: Flatten the 2D list into a single list of numbers
    flat_list = [item for sublist in lst for item in sublist]

    # Step 2: Handle empty list to avoid division by zero
    if not flat_list:
        return 0.0

    # Step 3: Calculate average (Sum / Count)
    return sum(flat_list) / len(flat_list)

nested_numbers = [[1, 2], [3, 4], [5, 6]]
result = average_nested(nested_numbers)

print(f"The average of {nested_numbers} is: {result}")
```

```python
A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

# Prove A · A_inv = I (Identity Matrix)
A_inv = np.linalg.inv(A)
I_result = np.dot(A, A_inv)
I_expected = np.eye(2)
proof1 = np.allclose(I_result, I_expected)
print(f"1. A * A_inv is Identity Matrix: {proof1}")
print(I_result)

# Prove AB != BA (Matrix multiplication is not commutative)
AB = np.dot(A, B)
BA = np.dot(B, A)
proof2 = not np.allclose(AB, BA)
print(f"\n2. AB is NOT equal to BA: {proof2}")
print("AB:\n", AB, "\nBA:\n", BA)

# Prove (AB).T = B.T · A.T (Transpose of a product)
lhs = (np.dot(A, B)).T
rhs = np.dot(B.T, A.T)
proof3 = np.allclose(lhs, rhs)
print(f"\n3. (AB).T == B.T * A.T: {proof3}")
```

### Experiment: How Fast is Numpy?

```python
import time

# Set the size of the arrays
size = 1_000_000

# Python List Implementation
# Create two lists of 1 million elements each
list1 = list(range(size))
list2 = list(range(size))

start_time = time.time()
# Element-wise addition using a list comprehension
list_result = [list1[i] + list2[i] for i in range(size)]
end_time = time.time()

python_time = end_time - start_time
print(f"Python List Addition Time: {python_time:.6f} seconds")

# NumPy Array Implementation
# Create two NumPy arrays of 1 million elements each
array1 = np.arange(size)
array2 = np.arange(size)

start_time = time.time()
# Element-wise addition using NumPy's vectorized operation
array_result = array1 + array2
end_time = time.time()

numpy_time = end_time - start_time
print(f"NumPy Array Addition Time: {numpy_time:.6f} seconds")

print(f"\nNumPy is approximately {python_time / numpy_time:.2f}x faster than Python lists!")
```

```python
# --- 1. Python List Implementation ---
size = 200 # Using 200 to keep the execution time reasonable
A_list = [[i for i in range(size)] for j in range(size)]
B_list = [[i for i in range(size)] for j in range(size)]

start_time = time.time()
# Manual nested loops (Triple for-loop)
result_list = [[0 for _ in range(size)] for _ in range(size)]
for i in range(size):
    for j in range(size):
        for k in range(size):
            result_list[i][j] += A_list[i][k] * B_list[k][j]
end_time = time.time()

python_time = end_time - start_time
print(f"Python List (200x200) Time: {python_time:.4f} seconds")

# --- 2. NumPy Array Implementation ---
size_np = 1000
A_np = np.random.rand(size_np, size_np)
B_np = np.random.rand(size_np, size_np)

start_time = time.time()
# Using np.dot() or the @ operator
result_np = np.dot(A_np, B_np)
end_time = time.time()

numpy_time = end_time - start_time
print(f"NumPy Array (1000x1000) Time: {numpy_time:.4f} seconds")
```
